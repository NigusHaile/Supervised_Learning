# Raccoon R-CNN Detection Pipeline

1. Split the raccoon dataset into **150 training images** and **50 test images**.
2. Generate positive and negative training ROIs using **Selective Search + IoU**.
3. Create a **hard negative test set** with 50 images containing no raccoon.
4. Use a pretrained CNN (**MobileNetV2**) as a fixed feature extractor.
5. Train an **SVM** on CNN features.
6. Implement inference: proposals → CNN features → SVM scores → thresholding → NMS.
7. Measure detection quality and computational time.
8. Search for a good accuracy/time trade-off.

## Utility functions

```python
def read_gt_boxes(image_path):

    image_path = Path(image_path)
    annot_path = orginal_annotations / f"{image_path.stem}.xml"
    contents = annot_path.read_text()
    soup = BeautifulSoup(contents, "html.parser")

    img_w = int(soup.find("width").string)
    img_h = int(soup.find("height").string)

    boxes = []
    for obj in soup.find_all("object"):
        label = obj.find("name").string.lower()
        if label != "raccoon":
            continue
        x1 = max(0, int(obj.find("xmin").string))
        y1 = max(0, int(obj.find("ymin").string))
        x2 = min(img_w - 1, int(obj.find("xmax").string))
        y2 = min(img_h - 1, int(obj.find("ymax").string))
        boxes.append((x1, y1, x2, y2))
    return boxes

def compute_iou(boxA, boxB):
    xA = max(boxA[0], boxB[0])
    yA = max(boxA[1], boxB[1])
    xB = min(boxA[2], boxB[2])
    yB = min(boxA[3], boxB[3])

    inter_w = max(0, xB - xA + 1)
    inter_h = max(0, yB - yA + 1)
    inter_area = inter_w * inter_h

    boxA_area = max(0, boxA[2] - boxA[0] + 1) * max(0, boxA[3] - boxA[1] + 1)
    boxB_area = max(0, boxB[2] - boxB[0] + 1) * max(0, boxB[3] - boxB[1] + 1)

    denom = float(boxA_area + boxB_area - inter_area)
    return inter_area / denom if denom > 0 else 0.0

def selective_search(image, mode="fast"):
    
    ss = cv2.ximgproc.segmentation.createSelectiveSearchSegmentation()
    ss.setBaseImage(image)
    if mode == "quality":
        ss.switchToSelectiveSearchQuality()
    else:
        ss.switchToSelectiveSearchFast()
    rects = ss.process()

    proposals = []
    H, W = image.shape[:2]
    for (x, y, w, h) in rects:
        x1, y1 = max(0, x), max(0, y)
        x2, y2 = min(W - 1, x + w), min(H - 1, y + h)
        if (x2 - x1) >= 10 and (y2 - y1) >= 10:
            proposals.append((x1, y1, x2, y2))
    return proposals

def non_max_suppression(boxes, scores, iou_threshold=0.3):
    
    if len(boxes) == 0:
        return []

    boxes = np.array(boxes, dtype=np.float32)
    scores = np.array(scores, dtype=np.float32)

    x1, y1, x2, y2 = boxes[:,0], boxes[:,1], boxes[:,2], boxes[:,3]
    areas = (x2 - x1 + 1) * (y2 - y1 + 1)
    order = scores.argsort()[::-1]

    keep = []
    while len(order) > 0:
        i = order[0]
        keep.append(i)

        xx1 = np.maximum(x1[i], x1[order[1:]])
        yy1 = np.maximum(y1[i], y1[order[1:]])
        xx2 = np.minimum(x2[i], x2[order[1:]])
        yy2 = np.minimum(y2[i], y2[order[1:]])

        w = np.maximum(0, xx2 - xx1 + 1)
        h = np.maximum(0, yy2 - yy1 + 1)
        inter = w * h
        iou = inter / (areas[i] + areas[order[1:]] - inter)

        inds = np.where(iou <= iou_threshold)[0]
        order = order[inds + 1]

    return keep
```

## Build training ROI dataset from Selective Search proposals

Labeling rule:

- Positive ROI: maximum IoU with a raccoon ground-truth box is `> 0.70`.
- Negative ROI: maximum IoU with every raccoon box is `< 0.05`.

 I used Selective Search object proposals, warp each candidate to CNN input size, then save labeled proposal crops.

```python
def clean_roi_folders():
    if roi_base.exists():
        shutil.rmtree(roi_base)
    positive_path.mkdir(parents=True, exist_ok=True)
    negative_path.mkdir(parents=True, exist_ok=True)

def save_roi(image, box, output_path):
    x1, y1, x2, y2 = box
    roi = image[y1:y2+1, x1:x2+1]
    if roi.size == 0:
        return False
    roi = cv2.resize(roi, INPUT_DIMS, interpolation=cv2.INTER_CUBIC)
    cv2.imwrite(str(output_path), roi)
    return True

def build_roi_dataset(force_rebuild=False):
    if force_rebuild:
        clean_roi_folders()

    existing_pos = len(list(positive_path.glob("*.png")))
    existing_neg = len(list(negative_path.glob("*.png")))
    if existing_pos > 0 and existing_neg > 0:
        print(f"ROI dataset already exists: {existing_pos} positive, {existing_neg} negative")
        return

    clean_roi_folders()
    total_pos = 0
    total_neg = 0

    for i, image_path in enumerate(train_images, 1):
        print(f"[ROI] {i}/{len(train_images)} {Path(image_path).name}")
        image = cv2.imread(image_path)
        if image is None:
            continue

        gt_boxes = read_gt_boxes(image_path)
        proposals = selective_search(image, mode="fast")[:MAX_PROPOSALS_TRAIN]

        pos_count = 0
        neg_count = 0

        for prop in proposals:
            ious = [compute_iou(gt, prop) for gt in gt_boxes]
            max_iou = max(ious) if len(ious) > 0 else 0.0

            if max_iou > POS_IOU_THRESHOLD and pos_count < MAX_POSITIVE_PER_IMAGE:
                out = positive_path / f"raccoon_{total_pos:06d}.png"
                if save_roi(image, prop, out):
                    total_pos += 1
                    pos_count += 1

            elif max_iou < NEG_IOU_THRESHOLD and neg_count < MAX_NEGATIVE_PER_IMAGE:
                out = negative_path / f"no_raccoon_{total_neg:06d}.png"
                if save_roi(image, prop, out):
                    total_neg += 1
                    neg_count += 1

            if pos_count >= MAX_POSITIVE_PER_IMAGE and neg_count >= MAX_NEGATIVE_PER_IMAGE:
                break

    print("Saved positive ROIs:", total_pos)
    print("Saved negative ROIs:", total_neg)

build_roi_dataset(force_rebuild=False)
print("Final positive ROIs:", len(list(POSITIVE_PATH.glob('*.png'))))
print("Final negative ROIs:", len(list(NEGATIVE_PATH.glob('*.png'))))
```

## Create a hard negative test set from no-raccoon images

```python
def create_hard_negative_test_set(force_rebuild=False):
    if force_rebuild and hard_test.exists():
        shutil.rmtree(hard_test)
    hard_test.mkdir(parents=True, exist_ok=True)

    existing = sorted(list(hard_test.glob("*.jpg")) + list(hard_test.glob("*.png")))
    if len(existing) >= test_size:
        print("Hard test set already exists:", len(existing))
        return existing[:test_size]

    count = len(existing)
    candidates = train_images.copy()
    random.shuffle(candidates)

    for image_path in candidates:
        if count >= test_size:
            break

        image = cv2.imread(image_path)
        if image is None:
            continue
        H, W = image.shape[:2]
        gt_boxes = read_gt_boxes(image_path)

        proposals = selective_search(image, mode="fast")[:MAX_PROPOSALS_TRAIN]
        proposals = sorted(proposals, key=lambda b: (b[2]-b[0])*(b[3]-b[1]), reverse=True)

        for box in proposals:
            x1, y1, x2, y2 = box
            bw, bh = x2 - x1 + 1, y2 - y1 + 1
            area_ratio = (bw * bh) / (W * H)

            max_iou = max([compute_iou(gt, box) for gt in gt_boxes]) if gt_boxes else 0
            if max_iou < NEG_IOU_THRESHOLD and 0.20 <= area_ratio <= 0.80:
                crop = image[y1:y2+1, x1:x2+1]
                if crop.size == 0:
                    continue
                # Save as an image, not as an ROI training example.
                out = hard_test / f"hard_no_raccoon_{count:03d}.jpg"
                cv2.imwrite(str(out), crop)
                count += 1
                break

    # Fallback: if selective-search crops are not enough, generate blurred background crops.
    while count < test_size:
        image_path = random.choice(train_images)
        image = cv2.imread(image_path)
        H, W = image.shape[:2]
        crop = cv2.GaussianBlur(image, (31, 31), 0)
        out = hard_test / f"hard_no_raccoon_{count:03d}.jpg"
        cv2.imwrite(str(out), crop)
        count += 1

    hard_paths = sorted(list(hard_test.glob("*.jpg")) + list(hard_test.glob("*.png")))[:test_size]
    print("Created hard negative images:", len(hard_paths))
    return hard_paths

hard_test_images = create_hard_negative_test_set(force_rebuild=False)
hard_test_images = [str(p) for p in hard_test_images]
hard_test_images[:3]
```

## Pretrained CNN feature extractor

I used **MobileNetV2 pretrained on ImageNet** because it is much faster than VGG/AlexNet while still producing strong semantic features. The classifier head is removed, and each ROI becomes a 1280-dimensional feature vector.

```python
# ImageNet preprocessing for MobileNetV2
preprocess = transforms.Compose([
    transforms.Resize(INPUT_DIMS),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225])
])

try:
    weights = models.MobileNet_V2_Weights.DEFAULT
    cnn = models.mobilenet_v2(weights=weights)
except Exception:
    cnn = models.mobilenet_v2(pretrained=True)

# Remove classifier; keep convolutional features + global pooling
feature_extractor = nn.Sequential(
    cnn.features,
    nn.AdaptiveAvgPool2d((1, 1)),
    nn.Flatten()
).to(device)
feature_extractor.eval()

@torch.no_grad()
def extract_features_from_paths(image_paths_, batch_size=64):
    all_features = []

    for start in range(0, len(image_paths_), batch_size):
        batch_paths = image_paths_[start:start+batch_size]
        tensors = []
        for p in batch_paths:
            img = Image.open(p).convert("RGB")
            tensors.append(preprocess(img))
        batch = torch.stack(tensors).to(device)
        feats = feature_extractor(batch)
        all_features.append(feats.cpu().numpy())

    return np.vstack(all_features)

@torch.no_grad()
def extract_features_from_rois_bgr(rois_bgr, batch_size=64):
    all_features = []
    if len(rois_bgr) == 0:
        return np.empty((0, 1280), dtype=np.float32)

    for start in range(0, len(rois_bgr), batch_size):
        batch_rois = rois_bgr[start:start+batch_size]
        tensors = []
        for roi in batch_rois:
            roi_rgb = cv2.cvtColor(roi, cv2.COLOR_BGR2RGB)
            img = Image.fromarray(roi_rgb)
            tensors.append(preprocess(img))
        batch = torch.stack(tensors).to(device)
        feats = feature_extractor(batch)
        all_features.append(feats.cpu().numpy())

    return np.vstack(all_features)
```

## Extract CNN features from positive/negative saved proposals

```python
def build_feature_dataset(force_rebuild=False):
    if features_path.exists() and not force_rebuild:
        data = np.load(features_path)
        X, y = data["X"], data["y"]
        print("Loaded features:", X.shape, y.shape)
        return X, y

    pos_paths = sorted([str(p) for p in positive_path.glob("*.png")])
    neg_paths = sorted([str(p) for p in negative_path.glob("*.png")])

    # Balance classes for stable SVM training
    n = min(len(pos_paths), len(neg_paths))
    random.shuffle(pos_paths)
    random.shuffle(neg_paths)
    pos_paths = pos_paths[:n]
    neg_paths = neg_paths[:n]

    print("Positive ROIs:", len(pos_paths))
    print("Negative ROIs:", len(neg_paths))

    X_pos = extract_features_from_paths(pos_paths, batch_size=64)
    X_neg = extract_features_from_paths(neg_paths, batch_size=64)

    X = np.vstack([X_pos, X_neg])
    y = np.array([1] * len(X_pos) + [0] * len(X_neg))

    features_path.parent.mkdir(parents=True, exist_ok=True)
    np.savez_compressed(features_path, X=X, y=y)
    return X, y

X, y = build_feature_dataset(force_rebuild=False)
print("Feature matrix:", X.shape)
print("Labels:", np.bincount(y))
```

## Train SVM classifier

```python
# Train/validation split over ROI features to verify proposal-level separability
X_train, X_val, y_train, y_val = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

svm = make_pipeline(
    StandardScaler(),
    LinearSVC(C=1.0, class_weight="balanced", max_iter=10000, random_state=42)
)

svm.fit(X_train, y_train)

val_pred = svm.predict(X_val)
print(classification_report(y_val, val_pred, target_names=["no_raccoon", "raccoon"]))

cm = confusion_matrix(y_val, val_pred)
disp = ConfusionMatrixDisplay(cm, display_labels=["no_raccoon", "raccoon"])
disp.plot()
plt.title("SVM validation confusion matrix on ROI features")
plt.show()

with open(svm_path, "wb") as f:
    pickle.dump(svm, f)
print("Saved SVM:", svm_path)
```

## R-CNN inference pipeline

For each test image:

1. Generate Selective Search proposals.
2. Crop and resize each proposal.
3. Extract MobileNetV2 features.
4. Score proposals using the SVM decision function.
5. Keep proposals above threshold.
6. Apply Non-Maximum Suppression.

```python
def rcnn_inference(
    image_path,
    svm_model,
    max_proposals=DEFAULT_MAX_PROPOSALS_INFER,
    score_threshold=DEFAULT_SCORE_THRESHOLD,
    nms_iou=NMS_IOU_THRESHOLD,
    search_mode="fast",
    return_image=True
):
    start_time = time.time()

    image = cv2.imread(str(image_path))
    if image is None:
        raise ValueError(f"Could not read image: {image_path}")

    proposals = selective_search(image, mode=search_mode)[:max_proposals]

    rois = []
    valid_boxes = []
    for (x1, y1, x2, y2) in proposals:
        roi = image[y1:y2+1, x1:x2+1]
        if roi.size == 0:
            continue
        roi = cv2.resize(roi, INPUT_DIMS, interpolation=cv2.INTER_CUBIC)
        rois.append(roi)
        valid_boxes.append((x1, y1, x2, y2))

    if len(rois) == 0:
        return image if return_image else None, [], time.time() - start_time

    feats = extract_features_from_rois_bgr(rois, batch_size=64)
    scores = svm_model.decision_function(feats)

    candidate_boxes = []
    candidate_scores = []
    for box, score in zip(valid_boxes, scores):
        if score >= score_threshold:
            candidate_boxes.append(box)
            candidate_scores.append(float(score))

    keep = non_max_suppression(candidate_boxes, candidate_scores, iou_threshold=nms_iou)

    detections = []
    for idx in keep:
        detections.append({
            "box": tuple(map(int, candidate_boxes[idx])),
            "score": float(candidate_scores[idx])
        })

    elapsed = time.time() - start_time
    return image if return_image else None, detections, elapsed

def draw_detections(image, detections, color=(0, 255, 0)):
    out = image.copy()
    for det in detections:
        x1, y1, x2, y2 = det["box"]
        score = det["score"]
        cv2.rectangle(out, (x1, y1), (x2, y2), color, 2)
        cv2.putText(out, f"raccoon {score:.2f}", (x1, max(20, y1-8)),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.55, color, 2)
    return out
```

```python
# Test multiple images (at least 6)

num_test_images = min(6, len(test_images))

fig, axes = plt.subplots(2, 3, figsize=(18, 12))
axes = axes.ravel()

for i in range(num_test_images):

    image, detections, elapsed = rcnn_inference(
        test_images[i],
        svm,
        max_proposals=DEFAULT_MAX_PROPOSALS_INFER,
        score_threshold=DEFAULT_SCORE_THRESHOLD
    )

    print(f"\nImage {i+1}")
    print("Detections:", detections)
    print("Inference Time:", elapsed)

    vis = draw_detections(image, detections)

    axes[i].imshow(cv2.cvtColor(vis, cv2.COLOR_BGR2RGB))
    axes[i].set_title(f"Image {i+1}")
    axes[i].axis("off")

# Hide unused subplots if fewer than 6 images
for j in range(num_test_images, len(axes)):
    axes[j].axis("off")

plt.suptitle("R-CNN Detection Results on Test Images", fontsize=18)
plt.tight_layout()
plt.show()
```

## Detection metrics on the 50 raccoon test images

For each test image:

- **True Positive image**: at least one predicted box has IoU ≥ 0.50 with any ground-truth raccoon box.
- **False Negative image**: no predicted box matches the raccoon.
- **False Positive boxes**: predicted boxes that do not match any raccoon.
- We also report average best IoU and average runtime per image.

```python
def evaluate_raccoon_test_set(
    test_image_paths,
    svm_model,
    max_proposals=DEFAULT_MAX_PROPOSALS_INFER,
    score_threshold=DEFAULT_SCORE_THRESHOLD,
    nms_iou=NMS_IOU_THRESHOLD,
    match_iou=0.50,
    verbose=False
):
    rows = []
    total_time = 0.0

    for i, image_path in enumerate(test_image_paths, 1):
        _, detections, elapsed = rcnn_inference(
            image_path,
            svm_model,
            max_proposals=max_proposals,
            score_threshold=score_threshold,
            nms_iou=nms_iou,
            return_image=False
        )
        total_time += elapsed

        gt_boxes = read_gt_boxes(image_path)
        det_boxes = [d["box"] for d in detections]
        det_scores = [d["score"] for d in detections]

        best_iou = 0.0
        matched = False
        fp_boxes = 0

        for det_box in det_boxes:
            ious = [compute_iou(det_box, gt) for gt in gt_boxes]
            max_iou = max(ious) if ious else 0.0
            best_iou = max(best_iou, max_iou)
            if max_iou >= match_iou:
                matched = True
            else:
                fp_boxes += 1

        rows.append({
            "image": Path(image_path).name,
            "num_gt": len(gt_boxes),
            "num_det": len(detections),
            "best_iou": best_iou,
            "matched": int(matched),
            "fp_boxes": fp_boxes,
            "time_sec": elapsed,
            "max_score": max(det_scores) if det_scores else np.nan
        })

        if verbose:
            print(f"{i}/{len(test_image_paths)} matched={matched} det={len(detections)} IoU={best_iou:.3f} time={elapsed:.2f}s")

    df = pd.DataFrame(rows)
    summary = {
        "num_images": len(df),
        "recall_image_level": df["matched"].mean(),
        "mean_best_iou": df["best_iou"].mean(),
        "avg_detections_per_image": df["num_det"].mean(),
        "avg_false_positive_boxes_per_image": df["fp_boxes"].mean(),
        "avg_time_sec_per_image": df["time_sec"].mean(),
        "total_time_sec": total_time
    }
    return df, summary

test_df, test_summary = evaluate_raccoon_test_set(
    test_images, svm,
    max_proposals=DEFAULT_MAX_PROPOSALS_INFER,
    score_threshold=DEFAULT_SCORE_THRESHOLD,
    verbose=True
)

print(json.dumps(test_summary, indent=2))
test_df.head()
```

## Performance on hard negative test set

For hard negative images, there is no raccoon. Any detection is a false positive.

Metric: **False positive image rate** = percentage of hard images with at least one detection.

```python
def evaluate_hard_test_set(
    hard_image_paths,
    svm_model,
    max_proposals=DEFAULT_MAX_PROPOSALS_INFER,
    score_threshold=DEFAULT_SCORE_THRESHOLD,
    nms_iou=NMS_IOU_THRESHOLD,
    verbose=False
):
    rows = []
    total_time = 0.0

    
    for i, image_path in enumerate(hard_image_paths, 1):
        _, detections, elapsed = rcnn_inference(
            image_path,
            svm_model,
            max_proposals=max_proposals,
            score_threshold=score_threshold,
            nms_iou=nms_iou,
            return_image=False
        )
        total_time += elapsed

        rows.append({
            "image": Path(image_path).name,
            "num_det": len(detections),
            "false_positive_image": int(len(detections) > 0),
            "time_sec": elapsed,
            "max_score": max([d["score"] for d in detections]) if detections else np.nan
        })

        if verbose:
            print(f"{i}/{len(hard_image_paths)} fp_image={len(detections)>0} det={len(detections)} time={elapsed:.2f}s")

    df = pd.DataFrame(rows)
    summary = {
        "num_images": len(df),
        "false_positive_image_rate": df["false_positive_image"].mean(),
        "avg_detections_per_image": df["num_det"].mean(),
        "avg_time_sec_per_image": df["time_sec"].mean(),
        "total_time_sec": total_time
    }
    return df, summary

hard_df, hard_summary = evaluate_hard_test_set(
    hard_test_images, svm,
    max_proposals=DEFAULT_MAX_PROPOSALS_INFER,
    score_threshold=DEFAULT_SCORE_THRESHOLD,
    verbose=True
)

print(json.dumps(hard_summary, indent=2))
hard_df.head()
```

## Tune speed/performance trade-off

R-CNN is slow because every proposal is forwarded through a CNN. The main practical trade-off parameters are:

- `max_proposals`: fewer proposals = faster, but may miss the object.
- `score_threshold`: higher threshold = fewer false positives, but may reduce recall.
- `nms_iou`: lower value removes more overlapping boxes.

```python
def run_tradeoff_experiment(
    proposal_values=(100, 300, 500, 1000),
    threshold_values=(0.0, 0.25, 0.50, 0.75),
    nms_iou=NMS_IOU_THRESHOLD
):
    results = []

    for max_props in proposal_values:
        for thr in threshold_values:
            print(f"Testing max_props={max_props}, threshold={thr}")

            _, pos_summary = evaluate_raccoon_test_set(
                test_images, svm,
                max_proposals=max_props,
                score_threshold=thr,
                nms_iou=nms_iou,
                verbose=False
            )

            _, neg_summary = evaluate_hard_test_set(
                hard_test_images, svm,
                max_proposals=max_props,
                score_threshold=thr,
                nms_iou=nms_iou,
                verbose=False
            )

            # Balanced score: high recall, low false positive image rate, low runtime.
            score = (
                pos_summary["recall_image_level"]
                - neg_summary["false_positive_image_rate"]
                - 0.02 * pos_summary["avg_time_sec_per_image"]
            )

            results.append({
                "max_proposals": max_props,
                "score_threshold": thr,
                "recall": pos_summary["recall_image_level"],
                "mean_best_iou": pos_summary["mean_best_iou"],
                "false_positive_image_rate_hard": neg_summary["false_positive_image_rate"],
                "avg_time_pos_sec": pos_summary["avg_time_sec_per_image"],
                "avg_time_hard_sec": neg_summary["avg_time_sec_per_image"],
                "tradeoff_score": score
            })

    return pd.DataFrame(results).sort_values("tradeoff_score", ascending=False)

tradeoff_df = run_tradeoff_experiment(
    proposal_values=(100, 300, 500, 1000),
    threshold_values=(0.0, 0.25, 0.50, 0.75)
)

tradeoff_df
```

## Plot trade-off results

## Notes 

R-CNN logic:

- **Selective Search** proposes class-agnostic object candidates.
- Each candidate ROI is warped to the CNN input size.
- A pretrained CNN extracts neural features.
- A binary SVM classifies each region as `raccoon` or `no_raccoon`.
- NMS removes multiple high-overlap detections for the same object.
- The trade-off experiment shows the effect of proposals and threshold on both detection quality and runtime.

Expected behavior:

- Increasing `max_proposals` usually improves recall but increases runtime.
- Increasing `score_threshold` usually reduces false positives but may reduce recall.
- Hard negative images reveal whether the detector fires on natural background or non-raccoon patterns.
